```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
m = 2.222 
c = 0.123 

X = np.arange(10,30,0.13)
y = m*X + c

print(len(X))

split = (len(X) * 0.8 ) # yo chai data lai split gareko.🇳🇵

X_train , y_train = X[:int(split)], y[:int(split)]
X_test , y_test = X[int(split):], y[int(split):]
```

```python
def plot_data(X_train = X_train , y_train = y_train , X_test = X_test , y_test = y_test , predictions=None):
    plt.scatter(X_train , y_train , c="r" , label="Training Data")
    plt.scatter(X_test , y_test , c="b" , label="Test Data")
    
    if predictions is not None:
        plt.scatter(X_test , predictions , c="g", label="Predicted Data")

plot_data()
```

```python
W = np.random.randn()/2
b = np.random.randn()/2

print(f"Initially W = {W} and bias = {b}")
```

```python
y_preds = X_test*W+b
plot_data(predictions=y_preds)
```

```python
def loss_function(y_actual , y_preds):
    return (1/len(y_actual) * np.sum(np.square((y_actual - y_preds))))
```

```python
epochs = 200 
lr = 0.000000001
train_loss_list = []
test_loss_list = []

for i in range(epochs):
    #forward pass
    train_preds = X_train *W +b
    train_loss = loss_function(y_train , train_preds)
    
    gradient_W = (-2/len(X_train))  * np.sum((y_train-train_preds) * X_train)
    gradient_b = (-2/len(X_train))  * np.sum((y_train-train_preds))
    
    W = W - lr*gradient_W
    b = b - lr*gradient_b 
    
    
    # Evaluate the loss on test set 
    
    test_preds = X_test * W+b
    test_loss = loss_function(y_test , test_preds)
    train_loss_list.append(train_loss)
    test_loss_list.append(test_loss)
    
    if (i+1)%10 == 0:
        print(f"Epochs {i+1} | Train Loss {train_loss:.4f} | Test Loss {test_loss:.4f}")
```

```python
plt.plot(range(epochs), train_loss_list, label="Training Loss")
plt.plot(range(epochs), test_loss_list, label="Testing Loss")
plt.title("Epochs vs Loss");
plt.xlabel("Epochs");
plt.ylabel("Loss")
plt.legend();
plt.show()
```
